```python
# Accept a string from the user and display it in reverse order

text = input("Enter a string: ")
reversed_text = text[::-1]
print("Reversed string:", reversed_text)
```

```python
# Count the vowels and consonants in a string (ignoring non-alphabetic characters)

text = input("Enter a string: ").lower()
vowels = "aeiou"
v_count = 0
c_count = 0

for char in text:
  if char.isalpha():
    if char in vowels:
      v_count += 1
    else:
      c_count += 1

print(f"Vowels: {v_count}")
print(f"Consonants: {c_count}")
```

```python
# Check whether a word or phrase is a palindrome, ignoring spaces and case

text = input("Enter a word or phrase: ")
# Remove spaces and convert to lowercase
cleaned = "".join(text.split()).lower()

if cleaned == cleaned[::-1]:
  print("The string is a palindrome.")
else:
  print("The string is not a palindrome.")
```

```python
# Accept a sentence and count its words, characters, and spaces

sentence = input("Enter a sentence: ")

characters = len(sentence)
spaces = sentence.count(" ")
words = len(sentence.split())

print(f"Characters: {characters}")
print(f"Spaces: {spaces}")
print(f"Words: {words}")
```

```python
# Count how many times each character occurs in a string

text = input("Enter a string: ")
freq = {}

for char in text:
  freq[char] = freq.get(char, 0) + 1

print("Character Frequencies:")
for char, count in freq.items():
  print(f"'{char}': {count}")
```

```python
# Accept five integers in a list and display their sum, average, maximum, and minimum

numbers = []
print("Enter 5 integers:")
for i in range(5):
  num = int(input(f"Integer {i+1}: "))
  numbers.append(num)

total_sum = sum(numbers)
average = total_sum / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print(f"Sum: {total_sum}")
print(f"Average: {average}")
print(f"Maximum: {maximum}")
print(f"Minimum: {minimum}")
```

```python
# Remove duplicate elements from a list while keeping the original order

original_list = [1, 2, 3, 2, 4, 1, 5, 3]
unique_list = []

for item in original_list:
  if item not in unique_list:
    unique_list.append(item)

print("Original List:", original_list)
print("List without duplicates:", unique_list)
```

```python
# Accept numbers from the user and display them in ascending and descending order

n = int(input("How many numbers do you want to enter? "))
numbers = []
for i in range(n):
  numbers.append(int(input(f"Enter number {i+1}: ")))

numbers.sort()
print("Ascending order:", numbers)

numbers.sort(reverse=True)
print("Descending order:", numbers)
```

```python
# Separate the even and odd numbers from a list into two lists

numbers = [10, 15, 22, 33, 40, 55, 60]
evens = []
odds = []

for num in numbers:
  if num % 2 == 0:
    evens.append(num)
  else:
    odds.append(num)

print("Original List:", numbers)
print("Even Numbers:", evens)
print("Odd Numbers:", odds)
```

```python
# Demonstrate append(), insert(), remove(), pop(), sort(), and reverse()

my_list = [30, 10, 50]
print("Initial list:", my_list)

my_list.append(40)  # Adds 40 to the end
print("After append(40):", my_list)

my_list.insert(1, 20)  # Inserts 20 at index 1
print("After insert(1, 20):", my_list)

my_list.remove(10)  # Removes the first occurrence of 10
print("After remove(10):", my_list)

popped_item = my_list.pop()  # Removes and returns the last item
print(f"After pop() [removed {popped_item}]:", my_list)

my_list.sort()  # Sorts the list in ascending order
print("After sort():", my_list)

my_list.reverse()  # Reverses the order of elements
print("After reverse():", my_list)
```

```python
# Count how many times an element appears in a tuple and find its first index

my_tuple = (1, 2, 3, 2, 4, 2, 5)
element = 2

count = my_tuple.count(element)
first_index = my_tuple.index(element)

print(f"Element {element} appears {count} times.")
print(f"First occurrence index of {element}: {first_index}")
```

```python
# Convert a list into a tuple, then convert the tuple back into a list

my_list = ["apple", "banana", "cherry"]
print("Original List:", my_list, type(my_list))

# List to Tuple
my_tuple = tuple(my_list)
print("Converted to Tuple:", my_tuple, type(my_tuple))

# Tuple back to List
converted_list = list(my_tuple)
print("Converted back to List:", converted_list, type(converted_list))
```

```python
# Perform union, intersection, difference, and symmetric difference on two sets

set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
print("Symmetric Difference:", set_a ^ set_b)
```

```python
# Accept a list of numbers and remove duplicate elements using a set

numbers = [1, 2, 2, 3, 4, 4, 5, 1]
print("Original List with duplicates:", numbers)

unique_numbers = list(set(numbers))
print("List after removing duplicates using a set:", unique_numbers)
```

```python
# Check whether one set is a subset or superset of another set

set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}

print("Set 1:", set1)
print("Set 2:", set2)

print("Is Set 1 a subset of Set 2?", set1.issubset(set2))
print("Is Set 2 a superset of Set 1?", set2.issuperset(set1))
```

```python
# Store five students' names and marks in a dictionary, display them, and calculate average marks

students = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 90, "Ethan": 88}

print("Student Marks:")
for name, marks in students.items():
  print(f"{name}: {marks}")

average_marks = sum(students.values()) / len(students)
print(f"Average Marks: {average_marks}")
```

```python
# Identify the student who has obtained the highest marks from a dictionary

students = {"Alice": 85, "Bob": 95, "Charlie": 78, "Diana": 90}

top_student = max(students, key=students.get)
print(
    f"The student with the highest marks is {top_student} with"
    f" {students[top_student]} marks."
)
```

```python
# Demonstrate keys(), values(), items(), get(), update(), and pop()

student = {"name": "John", "age": 20, "course": "Engineering"}

print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())

print("Get 'age':", student.get("age"))

student.update({"city": "Mumbai", "age": 21})
print("After update():", student)

removed_value = student.pop("course")
print(f"After pop('course') [removed {removed_value}]:", student)
```

```python
# Count the frequency of each word in a sentence, treating uppercase and lowercase as the same

sentence = (
    "Python is easy and Python is powerful and popular for data science"
).lower()
words = sentence.split()
word_count = {}

for word in words:
  word_count[word] = word_count.get(word, 0) + 1

print("Word Frequencies:")
for word, count in word_count.items():
  print(f"'{word}': {count}")
```
